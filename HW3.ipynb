```python
#Import packages
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Read file
aqs2023 = "/Users/gabrielagarcia/Desktop/Fall 2026/Geol 415/hw2/LA_AQS_2023.csv"
df = pd.read_csv(aqs2023)
```

```python
#creating subset of data pulling from data set and specific columns
df_O3 = df[(df['Parameter Name']=='Ozone') & (df['Duration Description']=='1 HOUR')]
df_O3.head()
```

```python
df_O3.columns
```

```python
#Change date from string to number
df_O3["Date (Local)"] = pd.to_datetime(df_O3["Date (Local)"])

```

```python
#Assign name to datasets
date = df_O3["Date (Local)"]
value = df_O3["Arithmetic Mean"]
```

```python
#Make scatterplot
plt.figure(figsize=(8, 6))
sns.scatterplot(x=date, y=value)
```

```python
#find highest value
highest_val = value.max()
print(highest_val)
```
